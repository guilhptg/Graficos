```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as colors


paises_df = pd.read_csv('Olympics_Country.csv')
jogos_df = pd.read_csv('Olympic_Games_Medal_Tally.csv')
resultados_altle_df = pd.read_csv('Olympic_Athlete_Event_Results.csv')
games_medal_df = pd.read_csv('Olympic_Games_Medal_Tally.csv')
games_df = pd.read_csv('Olympics_Games.csv')

display(games_medal_df)
display(games_df)
```

# Medalhas

```python
games_medal_2022_df = games_medal_df[games_medal_df['edition'] == '2022 Winter Olympics']
country = games_medal_2022_df[['country_noc']]
country_name = games_medal_2022_df[['country']]
gold = games_medal_2022_df[['gold']]
plt.figure(figsize=(18,5))
plt.plot(country['country_noc'], gold['gold'], marker='o')
plt.xlabel('Country')
plt.ylabel('Gold Medal')
plt.title('Gold Medal x Country')
plt.grid(True)
plt.show()
```

# Barras

```python

silver = games_medal_2022_df[['silver']]
bronze = games_medal_2022_df[['bronze']]

bar_plot = plt.figure(figsize=(18,5))
bar_plot = plt.bar(country['country_noc'], silver['silver'], label='Silver', color='gray')
bar_plot = plt.bar(country['country_noc'], bronze['bronze'], label='Bronze', color='red')
bar_plot = plt.bar(country['country_noc'], gold['gold'], label='Gold', color='yellow')
bar_plot = plt.bar_label(bar_plot,labels=country, padding=2, fontsize=9)
bar_plot = plt.xticks(ticks=country['country_noc'])
bar_plot = plt.xlabel('Countrys')

bar_plot = plt.ylabel('Medals')
bar_plot = plt.legend()
bar_plot = plt.grid(True)
bar_plot = plt.show()
```

# Pizza / Pie

```python
pie_plot = plt.figure(figsize=(18,5))
pie_plot = plt.plot(country['country_noc'], silver['silver'], color='gray', label='Siver', marker='o', scalex=True, linewidth=3, markersize=4)
pie_plot = plt.plot(country['country_noc'], gold['gold'], color='yellow', label='Gold', marker='o', scalex=True, linewidth=3, markersize=4)
pie_plot = plt.plot(country['country_noc'], bronze['bronze'], color='brown', label='Bronze', marker='o', scalex=True, linewidth=3, markersize=4, animated=True)
pie_plot = plt.ylim(-1, 20)
pie_plot = plt.title('Medalhas Olimpiadas - 2022')
pie_plot = plt.ylabel('Medalhas (2022)')
pie_plot = plt.legend()
pie_plot = plt.grid(True)
pie_plot = plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cm
import matplotlib.cbook as cbook
import matplotlib.colors as colors

N = 100
X, Y = np.mgrid[-3:3:complex(0, N), -2:2:complex(0, N)]

# A low hump with a spike coming out of the top right.  Needs to have
# z/colour axis on a log scale, so we see both hump and spike. A linear
# scale only shows the spike.
Z1 = np.exp(-X**2 - Y**2)
Z2 = np.exp(-(X * 10)**2 - (Y * 10)**2)
Z = Z1 + 50 * Z2

fig, ax = plt.subplots(2, 1)

pcm = ax[0].pcolor(X, Y, Z,
                   norm=colors.LogNorm(vmin=Z.min(), vmax=Z.max()),
                   cmap='PuBu_r', shading='auto')
fig.colorbar(pcm, ax=ax[0], extend='max')

pcm = ax[1].pcolor(X, Y, Z, cmap='PuBu_r', shading='auto')
fig.colorbar(pcm, ax=ax[1], extend='max')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery-nogrid')


# make data
x = [1, 2, 3, 4]
colors = plt.get_cmap('Blues')(np.linspace(0.2, 0.7, len(x)))

# plot
fig, ax = plt.subplots()
ax.pie(x, colors=colors, radius=3, center=(4, 4),
       wedgeprops={"linewidth": 4, "edgecolor": "white"}, frame=True)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
